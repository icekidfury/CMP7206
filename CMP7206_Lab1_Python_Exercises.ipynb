## 2. Input and arithmetic

Ask the user for the number of records before and after cleaning. Convert both inputs to integers. Print the number of records removed and the percentage retained, rounded to one decimal place. Assume the initial count is greater than zero.


```python
# Write your code here
before = int(input("Enter number of records before cleaning: "))
after = int(input("Enter number of records after cleaning: "))

removed = before - after
percent_retained = round((after / before) * 100, 1)

print("Records removed:", removed)
print("Percentage retained:", percent_retained, "%")
```

## 5. For loops

Given `daily_sales = [12, 15, 9, 20, 14]`, use a `for` loop to calculate the total without calling `sum()`. Print the total.


```python
# Write your code here
daily_sales = [12, 15, 9, 20, 14]

total = 0
for sale in daily_sales:
    total += sale

print("Total sales:", total)
```

## 6. While loops

```python
# Write your code here
remaining = 5

while remaining >= 1:
    print(remaining)
    remaining -= 1

print("Finished")
```

## 7. Functions

Write a function `average_score(scores)` that returns the average of a non-empty list of scores. Call it with `[70, 80, 90]` and print the result.


```python
# Write your code here
def average_score(scores):
    return sum(scores) / len(scores)

result = average_score([70, 80, 90])
print(result)
```

## 12. Nested data challenge

Given the records below, use a loop to print the name of each student whose score is at least 50. Then calculate and print the mean score. Use only concepts covered in this lab.

`students = [{"name": "Amina", "score": 72}, {"name": "Ben", "score": 45}, {"name": "Chen", "score": 83}]`


```python
# Write your code here
students = [{"name": "Amina", "score": 72},
            {"name": "Ben", "score": 45},
            {"name": "Chen", "score": 83}]

total = 0
for student in students:
    if student["score"] >= 50:
        print(student["name"])
    total += student["score"]

mean_score = total / len(students)
print("Mean score:", mean_score)
```
